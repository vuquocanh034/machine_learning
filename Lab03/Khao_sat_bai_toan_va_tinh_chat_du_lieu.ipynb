## 1. Bối cảnh bài toán (theo paper De Cock, 2011)

- Dữ liệu mô tả các giao dịch mua bán nhà ở riêng lẻ tại thành phố **Ames, Iowa, Hoa Kỳ**, thu thập từ **Văn phòng Đánh giá thuế (City Assessor's Office)** trong giai đoạn **2006-2010**.
- Tác giả đề xuất bộ dữ liệu này như một **phương án thay thế cho bộ dữ liệu Boston Housing** (quá cũ, ít thuộc tính) để làm đồ án cuối kỳ môn hồi quy cho sinh viên thống kê/khoa học dữ liệu.
- Bộ dữ liệu gốc có **2930 quan sát và 82 thuộc tính**; phiên bản đưa lên cuộc thi Kaggle được chia thành **train (1460 dòng, có sẵn SalePrice)** và **test (1459 dòng, ẩn SalePrice)**, cùng **81 thuộc tính** mô tả gần như mọi khía cạnh của ngôi nhà (diện tích, chất lượng, năm xây, vị trí, tiện ích...).
- **Bài toán:** Dự đoán **SalePrice** (giá bán nhà) dựa trên các thuộc tính đầu vào -> đây là bài toán **hồi quy (regression)**.
- Lý do bộ dữ liệu này phù hợp cho EDA là có cả biến số và biến phân loại, có giá trị thiếu tự nhiên và có một số mẫu bất thường được tác giả lưu ý trong paper.


```python
import pandas as pd
import numpy as np
pd.set_option('display.max_columns', 15)

df = pd.read_csv('train.csv')
print("Kích thước dữ liệu (train):", df.shape)
df.head()
```

## 2. Tính chất dữ liệu tổng quan

```python
print("Số dòng:", df.shape[0])
print("Số cột:", df.shape[1])
print()
df.info(verbose=False)
```

```python
dtype_counts = df.dtypes.value_counts()
print("Phân loại kiểu dữ liệu các cột:\n", dtype_counts)

numeric_cols = df.select_dtypes(include=[np.number]).columns.tolist()
categorical_cols = df.select_dtypes(include=['object']).columns.tolist()
print(f"\nSố biến số (numeric): {len(numeric_cols)}")
print(f"Số biến phân loại (categorical): {len(categorical_cols)}")
```

**Nhận xét:** Dữ liệu gồm 1460 quan sát, 81 cột (bao gồm `Id` và `SalePrice`).
- 38 biến dạng số (numeric: int64/float64), trong đó có cả biến định lượng thực sự (diện tích, giá) và biến có bản chất thứ tự/danh mục nhưng được lưu dưới dạng số (VD: `OverallQual`, `MSSubClass`).
- 43 biến dạng chuỗi (object/categorical), mô tả các đặc điểm danh mục của ngôi nhà (vị trí, kiểu nhà, vật liệu, chất lượng theo thang chữ...).


## 2.1. Phân biệt kiểu biến theo ý nghĩa dữ liệu

Ngoài kiểu dữ liệu mà Python đọc được (`numeric` hoặc `object`), cần nhìn vào ý nghĩa của từng biến. Theo paper của De Cock, bộ dữ liệu Ames gồm các nhóm biến **nominal, ordinal, discrete và continuous**. Vì vậy, có những biến được lưu bằng số nhưng không nên hiểu đơn giản là biến định lượng liên tục.

Ví dụ, `OverallQual` là mức đánh giá chất lượng theo thứ tự, còn `MSSubClass` là mã nhóm nhà. Ngược lại, `GrLivArea`, `TotalBsmtSF` và `LotArea` là các biến định lượng liên tục.


```python
semantic_examples = pd.DataFrame({
    'Nhóm biến': ['Định danh', 'Mục tiêu', 'Định lượng liên tục', 'Định lượng rời rạc', 'Thứ bậc', 'Phân loại'],
    'Ví dụ': [
        'Id',
        'SalePrice',
        'GrLivArea, TotalBsmtSF, LotArea',
        'BedroomAbvGr, FullBath, GarageCars',
        'OverallQual, OverallCond, KitchenQual',
        'Neighborhood, HouseStyle, SaleCondition'
    ]
})
semantic_examples
```

**Nhận xét bổ sung:** Như vậy, khi phân tích dữ liệu không nên chỉ dựa vào `dtype`. Cần kết hợp cả kiểu lưu trữ và ý nghĩa của biến để chọn cách phân tích phù hợp. Phần này giúp làm rõ hơn đặc điểm dữ liệu Ames được nêu trong paper.

## 3. Biến mục tiêu (SalePrice)

```python
print(df['SalePrice'].describe())
print("\nĐộ lệch (skewness):", round(df['SalePrice'].skew(), 3))
```

**Nhận xét:** `SalePrice` dao động từ **$34,900** đến **$755,000**, trung bình khoảng **$180,921**.
Độ lệch (skewness) = **1.88** > 0 cho thấy phân phối **lệch phải khá rõ**. Phần lớn nhà có giá ở mức trung bình đến thấp, trong khi một số ít nhà có giá rất cao kéo đuôi phân phối về phía phải. Phần này sẽ được phân tích kỹ hơn ở phần đơn biến (Thành viên 2).
